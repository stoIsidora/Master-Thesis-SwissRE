```python
import csv
import pandas as pd
import numpy as np
import matplotlib.colors as mcolors
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('thesis__data.csv', usecols = ['YEAR', 'AGE', 'MORTSTAT', 'MORTDODY', 'VIG10DMIN'])
```

```python
# remove what has to be removed
df = df[(df['MORTSTAT'] == 1) | (df['MORTSTAT'] == 2)]

# define year of birth
df['YEAR_BIRTH'] = df['YEAR'] - df['AGE']
```

```python
# computes the average number of people at each age over years
def avr_P_i(df, years, age_min, age_max):
    
    standardization_df = pd.DataFrame(index=years, columns=range(age_min, age_max + 1))
    
    for year in years:
        df_use = df[(df['MORTDODY'] >= year) & (df['YEAR'] <= year)].copy()

        # define age of these people in year 
        df_use['AGE_year'] = year - df_use['YEAR_BIRTH']
        df_use = df_use[df_use['AGE_year'] <= age_max]

        # count occurrences of each age for the year
        age_count = df_use['AGE_year'].value_counts().reindex(range(age_min, age_max + 1), fill_value=0)

        # assign the counts to the standardization DataFrame
        standardization_df.loc[year] = age_count.values
        
    mean_count = standardization_df.mean()
    
    return mean_count 
```

```python
# pre-selection
age_min = 18
age_max = 80
```

```python
# VIG10DMIN
```

```python
# fix universe 
df_vigmin = df.copy()
df_vigmin = df_vigmin[(df_vigmin['VIG10DMIN']<=996)]

df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 1 if 0 < x <= 10 else x)
df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 2 if 10 < x <= 20 else x)
df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 3 if 20 < x <= 30 else x)
df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 4 if 30 < x <= 40 else x)
df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 5 if 40 < x <= 50 else x)
df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 6 if 50 < x <= 60 else x)
df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 7 if 60 < x <= 90 else x)
df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 8 if 90 < x <= 120 else x)
df_vigmin['VIG10DMIN'] = df_vigmin['VIG10DMIN'].apply(lambda x: 9 if 120 < x   else x)
```

```python
# get populations
populations = sorted(df_vigmin['VIG10DMIN'].unique())

# initialize DataFrames to store SMR (standardized mortality rates) for each year
years = list(range(df_vigmin['YEAR'].min(), df_vigmin['YEAR'].max()+2))
smr_vigmin = pd.DataFrame(index=years, columns=populations)  

# track agining of population
age_avr = pd.DataFrame(index=years, columns=populations)

mean_count = avr_P_i(df_vigmin, years, age_min, age_max)

for year in years:
    # people who were alive at the beginning of year i  
    df_year = df_vigmin[(df_vigmin['MORTDODY'] >= year) & (df_vigmin['YEAR'] <= year)].copy()

    # define age of these people in year i
    df_year['AGE_year'] = year - df_year['YEAR_BIRTH']
    df_year = df_year[df_year['AGE_year'] <= age_max]
    
    # age counts (used for standardization)
    age_counts = []
    for age in range(age_min, age_max + 1):
        age_counts.append(df_year[df_year['AGE_year'] == age].shape[0])
    
    for population in populations:
        # work with one population at the time
        df_year_population  = df_year[df_year['VIG10DMIN'] == population]
        
        age_avr.at[year, population] = (df_year_population['AGE_year'].mean())
        
        # counts:
        count_all = []
        count_dead = []
        for age in range(age_min, age_max + 1):
            total_count = df_year_population[df_year_population['AGE_year'] == age].shape[0]
            dead_count = df_year_population[(df_year_population['MORTDODY'] == year) & (df_year_population['AGE_year'] == age)].shape[0]
            
            count_all.append(total_count)
            count_dead.append(dead_count)
        
        mr_per_age = [dead / total if total > 0 else 0 for dead, total in zip(count_dead, count_all)]
        
        smr_vigmin.at[year, population] = sum([a*b for a,b in zip(mr_per_age, mean_count)])/sum(mean_count)       
```

```python
# analyze aging effect
plt.figure(figsize=(6, 5))
labels = {0: '0', 1: '1-10', 2: '11-20', 3: '21-30', 4: '31-40', 5: '41-50', 6: '51-60', 7: '61-90', 8: '91-120', 9: '>120'}
colors = {0: '#0D47A1', 1: '#42A5F5', 2: '#90CAF9', 3: '#6FA86E', 4: '#1E7C44',  
          5: '#7DD3FC', 6: '#D8B4FE', 7: '#E879F9', 8: '#E06666', 9: '#D84315'}

for population in populations:
    plt.plot(age_avr.index, age_avr[population], marker='o', label=labels[population], color=colors[population])

plt.title('Agging effect')
plt.xlabel('year')
plt.ylabel('avarage age (of alive members of population)')
plt.xticks(smr_vigmin.index, rotation=90)  
plt.legend(title='average duration of\nvigorous activity')
plt.grid()
plt.tight_layout()
plt.savefig('agging_vigmin.jpg', format='jpg')
plt.show()
```

```python
plt.figure(figsize=(6, 5))
for population in populations:
    plt.plot(smr_vigmin.index, smr_vigmin[population], marker='o', label=labels[population], color=colors[population])

plt.xlabel('year')
plt.ylabel('Age Standardized Mortality Rate (ASMR)')
plt.xticks(smr_vigmin.index, rotation=90)  
plt.legend(title='average duration of\nvigorous activity')
plt.grid()
plt.tight_layout()
plt.savefig('smr_vigmin.jpg', format='jpg')
plt.show()
```

```python
# compute average per population
smr_mean = smr_vigmin.mean()

# define custom labels for x-axis
label_list = [labels[i] for i in range(len(labels))]
colors_list = [colors[i] for i in range(len(colors))]
```

```python
# --- plot ---
plt.figure(figsize=(5, 3.2))

# loop through the points and color them
for i, (x, y) in enumerate(zip(smr_mean.index, smr_mean.values)):
    plt.scatter(x, y, color=colors_list[i], s=100)
    
plt.title('Population average ASMR: Average duration of vigorous activity ', fontsize=10)
plt.xlabel('populations', fontsize=9)
plt.ylabel('mean ASMR', fontsize=9)
plt.xticks(smr_mean.index, label_list, rotation=0, fontsize=8)  
plt.yticks(fontsize=8)
plt.tight_layout()  
plt.grid(True)  
plt.savefig('avr_amr_vigmin.jpg', format='jpg')
plt.show()
```
